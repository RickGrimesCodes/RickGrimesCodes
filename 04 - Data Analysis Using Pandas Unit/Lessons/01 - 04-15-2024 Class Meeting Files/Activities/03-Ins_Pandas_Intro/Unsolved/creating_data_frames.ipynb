```python
# Dependencies
import pandas as pd # allows for comands as pd.function() as opposed to pandas.function()
```

```python
# We can create a Pandas Series from a raw list
schoollist = ["UCLA", "UC Berkeley", "UC Irvine", "University of Central Florida", "Rutgers University"]

# Converts the python list to a pandas series
dataSeries = pd.Series(schoollist)

dataSeries
```

```python
# Convert a list of dictionaries into a DataFrame
stateAbbrevs = [
    {"STATE": "New Jersey", "ABBREVIATION": "NJ"},
    {"STATE": "New Jersey", "ABBREVIATION": "NY"}
]
# use pd.DataFrame() to convert this list of dicts to a dataframe
stateDF = pd.DataFrame(stateAbbrevs)

stateDF
```
