```python
# Import Dependencies
import pandas as pd
from pathlib import Path
```

```python
# Create a path to the csv and read it into a Pandas DataFrame
movieDF = pd.read_csv("resources/movie_scores.csv")
movieDF.head()
```

```python
# Figure out the minimum and maximum IMDB user vote count
print(f"Maximum IMDB User Votes: {movieDF['IMDB_user_vote_count'].max()}")
print(f"Manimum IMDB User Votes: {movieDF['IMDB_user_vote_count'].min()}")
```

```python
# Create bins in which to place values based upon IMDB vote count
bins = [0, 2499, 4999, 9999, 14999, 19999, 29999, 49999, 99999, 350000]

# Create labels for these bins
group_labels = ["0 to 2.4k", "2.5k to 4.9k", "5k to 9k", "10k to 14k", "15k to 19k", "20k to 29k",
                "30k to 49k", "50k to 99k", "100k to 350k"]
```

```python
# Slice the data and place it into bins
pd.cut(movieDF["IMDB_user_vote_count"], bins, labels=group_labels, include_lowest=True)
movieDF.head
```

```python
# Place the data series into a new column inside of the DataFrame
movieDF["IMDB User Votes Group"] = pd.cut(movieDF["IMDB_user_vote_count"], bins, labels=group_labels, include_lowest=True)
movieDF.head()
```
