```python
# Import dependencies
import pandas as pd
from pathlib import Path
```

```python
# Store filepath in a variable and the data into a Pandas DataFrame
veteransDF = pd.read_csv("Resources/veterans.csv")
veteransDF.head()
```

```python
# Get the average percentage of veterans living within 75 miles of a cemetery
veteransDF["Percentage"].mean()
```

```python
# Get the columns.
veteransDF.columns
```

```python
# Converting the "Percentage" column to floats
    # for example, the first value (percentage) - 72%
    # We need to get rid of the percent symbol
    # Then convert the number to a numerical data type (int or float)
    # Last, calculate the mean of the series (column)
veteransDF["Percentage"] = veteransDF["Percentage"].str.replace("%","")
veteransDF["Percentage"] = veteransDF["Percentage"].astype('float')
veteransDF.head()

# Finding the average percentage of veterans living within 75 miles of a cemetery
veteransDF["Percentage"].mean()
```
