```python
# Dependencies
import pandas as pd
from pathlib import Path
```

```python
# Load in file
# Store filepath in a variable
movie_file = Path("Resources/movie_scores.csv")
```

```python
# Read and display the CSV with Pandas
movieDF = pd.read_csv(movie_file)
movieDF.head()
```

```python
# List all the columns in the table
# .columns to get the list
for col in movieDF.columns:
    print(col)
```

```python
# We only want IMDb data, so create a new table that takes the Film and all the columns relating to IMDB
columnList = ["FILM", "IMDB", "IMDB_norm", "IMDB_norm_round", "IMDB_user_vote_count"]
imdbDF = movieDF[columnList]
imdbDF.head()
```

```python
# We only like good movies, so find those that scored over 7, and ignore the norm rating
columnlist = ["FILM", "IMDB", "IMDB_user_vote_count"]
imdbDF = movieDF.loc[
    movieDF["IMDB"] > 7, columnList
]
imdbDF.head()
```

```python
# Find less popular movies--i.e., those with fewer than 20K votes
lessPopularDF = imdbDF.loc[
    imdbDF["IMDB_user_vote_count"] < 20000
]
lessPopularDF.head()
```

```python
# Finally, export this file to a spread so we can keep track of out new future watch list without the index
lessPopularDF.to_excel("movieWatchList.xlsx", index=False)
```
