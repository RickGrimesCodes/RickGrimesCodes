# Geoapify Places Demo

```python
# Dependencies
import requests
import json

# Import the API key
from config import geoapify_key
```

```python
# Set the geographical coordinates for Sydney, Australia
"""
City: Sydney, Australia
Latitude: -33.8698439
Longitude: 151.2082848
"""
latitude = -33.8698439
longitude = 151.2082848

# Set the parameters for the type of place
categories = "catering.restaurant"
conditions = "vegetarian"
radius = 8000

# Set the parameters for the type of search
filters = f"circle:{longitude},{latitude},{radius}"
bias = f"proximity:{longitude},{latitude}"
limit = 20

# set up a parameters dictionary
params = {
    "categories": categories,
    "conditions": conditions,
    "filter": filters,
    "bias": bias,
    "limit": limit,
    "apiKey": geoapify_key
}


# Set base URL
base_url = "https://api.geoapify.com/v2/places"

# run a request using our params dictionary
response = requests.get(base_url, params=params)
```

```python
# print the response url, avoid doing for public github repos in order to avoid exposing key
print(response.url)
```

```python
# convert response to json
response = response.json()

# Print the json (pretty printed)
from pprint import pprint
pprint(response)
```

```python
# Print the name and address of the first restaurant that appears
firstRestaurant = response["features"][0]["properties"]["name"]
address = response["features"][0]["properties"]["address_line2"]

print(firstRestaurant)
print(address)
```
