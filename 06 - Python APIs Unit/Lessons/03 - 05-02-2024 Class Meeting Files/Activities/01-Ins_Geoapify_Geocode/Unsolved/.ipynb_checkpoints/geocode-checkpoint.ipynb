# Geoapify Geocode Demo

```python
# Dependencies
import requests
import json

# Import the API key
from config import geoapify_key
```

```python
# Target city
target_city = "Elkin, United States"

# Build the endpoint URL
target_url = f"https://api.geoapify.com/v1/geocode/search?text={target_city}&format=json&apiKey={geoapify_key}"
```

```python
# Run a request to endpoint and convert result to json
response = requests.get(target_url).json()

# Print the json
```

```python
# Print the json (pretty printed)
from pprint import pprint
pprint(response)
```

```python
# Extract latitude and longitude
latitude = response["results"][0]['lat']
longitude = response["results"][0]['lon']


# Print the latitude and longitude
print(f"city: {target_city}")
print(f"Latitude: {latitude}")
print(f"Longitude: {longitude}")
```
