```python
# Dependencies
import json
import requests
from config import api_key
```

```python
# Save config information
url = "http://api.openweathermap.org/data/2.5/weather?"
city = "Elkin"

# Build query URL - http://api.openweathermap.org/data/2.5/weather?appid=3336cc4cb5ad77197707428d8e8b10de&q=Charlotte
queryURL = url + "appid=" + api_key + "&q=" + city
# or
queryURL = f"{url}appid={api_key}&q={city}"
queryURL
```

```python
# Get weather data
response = requests.get(queryURL).json()

# Get the temperature from the response
print(f"The weather API responded with: {response}")
```
