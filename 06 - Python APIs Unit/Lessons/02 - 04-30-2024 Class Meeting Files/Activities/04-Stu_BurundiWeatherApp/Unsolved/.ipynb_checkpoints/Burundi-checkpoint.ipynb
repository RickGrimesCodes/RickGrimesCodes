```python
# Dependencies
import requests
from config import api_key

# Save config information.
url = "http://api.openweathermap.org/data/2.5/weather?"
city = "Bujumbura"
units = 'metric' # celsisus = 'metric', fahrenheit = 'imperial'
```

```python
# Build query URL and request your results in Celsius - add &units= property
queryURL = f"{url}&appid={api_key}&q={city}&units={units}"
#queryURL

# Get weather data
weatherJSON = requests.get(queryURL).json()
```

```python
# Get temperature from JSON response = ['main']['temp']
currentTemperature = weatherJSON['main']['temp']
```

```python
# Report temperature
print(f"The temperature in Bujumbura is {currentTemperature} C.")
```

```python
# BONUS

# use list of units
# celsius = 'metric', fahrenheit = 'imperial'
units = ['metric', 'imperial']

# set up list to hold two different temperatures
temperatures = []

# loop through the list of units and append them to temperatures list
for unit in units:

    # Build query URL based on current element in units
    queryURL = f"{url}&appid={api_key}&q={city}&units={unit}"

    # Get weather data
    weatherJSON = requests.get(queryURL).json()

    # Get temperature from JSON response
    currentTemperature = weatherJSON['main']['temp']

    temperatures.append(currentTemperature)


# Report temperatures by accessing each element in the list
print(f"The temperature in Bujumbura is {temperatures[0]}C or {temperatures[1]}F.")
```
