```python
# Dependencies
import requests
import json
```

```python
# Base URL for GET requests to retrieve number/date facts
url = "http://numbersapi.com/"
```

```python
# Ask the user what kind of data they would like to search for
# all but the trivia facts have the type of fact at the end of the url
    # ex: math fact: http://numbersapi.com/29/math
searchType = input("What type of data would you like to search for? [Trivia, Math, Date, or Year] ")
```

```python
# Create code to return a number fact
# If the kind of search is "date" take in two numbers
if searchType.lower() == "date":
    
    # Collect the month to search for
    month = input("Enter the month to search for (1-12): ")
    
    # Collect the day to search for
    day = input("Enter the day to search for (1-31): ")
    
    # Make an API call to the "date" API and convert response object to JSON
    # ex: date fact: http://numbersapi.com/4/29/date
    response = requests.get(f"{url}{month}/{day}/date?json").json()
    
    # Print the fact stored within the response - stored in the "text" key
    print(response["text"])

# If the kind of search is anything but "date" then take one number
else:
    # Collect the number to search for
    number = input("What number would you like to search for? ")
    # Make an API call to the API and convert response object to JSON
    # ex: math fact: http://numbersapi.com/29/math
    response = requests.get(f"{url}{number}/{searchType.lower()}?json").json()
     # Print the fact stored within the response - stored in the "text" key
    print(response["text"])
```
