```python
# Dependencies
import requests
from config import api_key

url = "http://www.omdbapi.com/?apikey=" + api_key + "&t="

movies = ["Interstellar", "Oppenheimer", "Tron Legacy"]

movieResponses = []

# Make a request to the OMDb API for each movie in the list.
for movie in movies:

    movieInfo = requests.get(url+movie).json()
    # Print the director of each movie
    print(f"The director of {movie} is {movieInfo['Director']}.")
    # Save the responses in another list
    movieResponses.append(movieInfo)
```
