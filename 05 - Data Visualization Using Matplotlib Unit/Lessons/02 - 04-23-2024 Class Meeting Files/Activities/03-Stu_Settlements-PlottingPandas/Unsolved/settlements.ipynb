```python
# Dependencies
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Read CSV
settlementDF = pd.read_csv("Resources/union_settlements_1995.csv")
settlementDF.head()
```

```python
# List columns
for columns in settlementDF.columns:
    print(columns)
```

```python
# Get total settlements by union
# use the "UNION" column to get value counts
unionSettlements = settlementDF["UNION"].value_counts()
unionSettlements
```

```python
unionSettlements.index
```

```python
# Configure plot, figsize, title, and axis labels
# use .plot on the unionSettlements series
figure1 = unionSettlements.plot(kind="bar", color="red", xlabel="Union", ylabel="Settlements",
                      title="Major Collective Bargaining Settlements (1995)")
# Configure x-tick rotation
figure1.set_xticklabels(unionSettlements.index, rotation=45, wrap=True, rotation_mode="anchor", ha="right")

# Show plot
plt.show()
```
