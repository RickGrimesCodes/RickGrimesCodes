```python
# Dependencies
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Read CSV
unemployed_data_one = pd.read_csv("../Resources/unemployment_2010-2015.csv")
unemployed_data_two = pd.read_csv("../Resources/unemployment_2016-2020.csv")

# Merge our two data frames together
combinedDF = pd.merge(unemployed_data_one, unemployed_data_two, on="Country Name")
combinedDF.head()
```

```python
# Delete the duplicate 'Country Code' column and rename the first one back to 'Country Code'
del combinedDF["Country Code_y"]
combinedDF = combinedDF.rename(columns={
    "Country Code_x": "Country Code"
})

combinedDF.head()
```

```python
# Set the 'Country Code' to be our index for easy referencing of rows
combinedDF = combinedDF.set_index("Country Code")
combinedDF.head()
```

```python
years = [str(year) for year in range(2010, 2021)]
years
```

```python
# Collect the mean unemployment rates for the world
years = [str(year) for year in range(2010, 2021)]
combinedYearsDF = combinedDF[years]
avgYearUnemployment = combinedYearsDF.mean()

# Collect the years where data was collected
# either use the years variable or use avgYearUnemployment.keys()
```

```python
# Plot the world average as a line chart
world_avg, = plt.plot(years, avgYearUnemployment, color="b", label="World Average")

# Plot the unemployment values for a single country
usa, = plt.plot(years, combinedDF.loc["USA", years], color="green", label="US Average")

# Create a legend for our chart
plt.legend(handles=[world_avg, usa], loc="best")
# Show the chart
plt.show()
```
