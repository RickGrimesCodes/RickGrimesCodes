```python
# Import Dependencies
import matplotlib.pyplot as plt
import pandas as pd
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
# Import our data into pandas from CSV
accidentsDF = pd.read_csv("../Resources/accidents.csv")
accidentsDF.head()
```

```python
# Create a group based on the values in the 'FUNC_SYSNAME' column
# 'FUNC_SYSNAME' stores the type of road the accident occurred

# Count how many times each road type appears in our group
countRoadTypes = accidentsDF.groupby("FUNC_SYSNAME").count()["STATE"]
countRoadTypes
```

```python
# Create a bar chart based off of the group series from before
roadTypeChart = countRoadTypes.plot(kind="bar", figsize=(8, 6))

# Set the xlabel and ylabel using class methods
roadTypeChart.set_xlabel("Road Type")
roadTypeChart.set_ylabel("# of Accidents")

plt.tight_layout()

plt.show()
```
