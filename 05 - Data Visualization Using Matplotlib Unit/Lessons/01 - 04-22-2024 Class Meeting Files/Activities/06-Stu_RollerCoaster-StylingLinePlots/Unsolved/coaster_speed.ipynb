```python
# Import Dependencies
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Create the X and Y axis lists
time = np.arange(0,130,10)
danger_drop_speeds = [9, 8, 90, 85, 80, 70, 70, 65, 55, 60, 70, 65, 50]
railgun_speeds = [75, 70, 60, 65, 60, 45, 55, 50, 40, 40, 35, 35, 30]
```

```python
# Plot the charts and apply some styling
dangerDrop, = plt.plot(time, danger_drop_speeds, color="red", label="Danger Drop")
railGun, = plt.plot(time, railgun_speeds, color="blue", label="RailGun")

plt.title("Coaster Speed Over Time")
plt.xlabel("Coaswter Runtime (Seconds)")
plt.ylabel("Coaster Speed (MPH)")
plt.grid()
plt.legend(handles=[dangerDrop, railGun], loc="best")
plt.xlim(0, 120)
plt.ylim(5, 97)
plt.show()
           
```
