```python
# Dependencies
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Set x axis to numerical value for month
zAxis =np.arange(1,13,1)
zAxis
```

```python
# Average weather temp
points = [39, 42, 51, 62, 72, 82, 86, 84, 77, 65, 55, 44]
```

```python
# Plot the line
plt.plot(zAxis, points)
plt.show()
```

```python
# Convert to Celsius C = (F-32) * 0.56
cTemps = [(fTemp-32) * 0.56 for fTemp in points]
cTemps
```

```python
# Plot using Celsius
plt.plot(zAxis, cTemps)
plt.show
```

```python
# Plot both on the same chart
plt.plot(zAxis, cTemps)
plt.plot(zAxis, points)
plt.show()
```
