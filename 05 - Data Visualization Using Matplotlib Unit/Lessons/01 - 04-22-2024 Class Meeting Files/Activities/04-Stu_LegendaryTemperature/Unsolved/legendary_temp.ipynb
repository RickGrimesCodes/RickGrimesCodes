```python
# Dependencies
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Set x axis to numerical value for month
xAxis = np.arange(1, 13, 1)
xAxis
```

```python
# Avearge weather temp
points_F = [39, 42, 51, 62, 72, 82, 86, 84, 77, 65, 55, 44]
```

```python
# Convert to Celsius C = (F-32) * 0.56
points_C = [(fTemp-32) * 0.56 for fTemp in points_F]
points_C
```

```python
# Create a handle for each plot
fahrenheit, = plt.plot(xAxis, points_F, marker="+", color="blue", label="Fahrenheit")
celsius, = plt.plot(xAxis, points_C, marker="s", color="red", label="Celsius")
```
